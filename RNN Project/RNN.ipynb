```python
# Data cleaning and preprocessing

import re
import nltk

nltk.download('stopwords')

'''
# If we don't want one word in stopwords, we use

all_stopwords = stopwords.words('english')
all_stopwords.remove('not')  # Here we removed the word 'not'
'''
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
# Text preprocessing Step 1

corpus = []
for i in range(0 , len(df)):
  review = df["title"].iloc[i]
  review = re.sub('[^a-zA-Z]' , ' ' , str(review))
  review = review.lower()
  review = review.split()
  review = [ps.stem(word) for word in review if not word in stopwords.words('english')]
  review = ' '.join(review)
  corpus.append(review)
print(review)
```

```python
corpus
```

```python
# Text preprocessing Step 2

from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 2500)
x = cv.fit_transform(corpus).toarray()
y = df.iloc[:, -1].values

# or y = df["label"]
```

```python
x.shape
```

```python
# Train Test Split

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x , y , train_size=0.8 , random_state = 42)
```

```python
from sklearn.naive_bayes import MultinomialNB      # Naive bayes

NB_model = MultinomialNB().fit(x_train , y_train)

# Prediction

y_pred = NB_model.predict(x_test)

from sklearn.metrics import accuracy_score, cohen_kappa_score, classification_report

acc_score = accuracy_score(y_test , y_pred)
print(acc_score)

kappa_score = cohen_kappa_score(y_test , y_pred)
print(kappa_score)
```

# **RNN**

```python
# Apply RNN

from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding , SimpleRNN , Dense

#Embedding - to convert text to vectors, we already used count vectorizer so here we no need to use Embedding
```

```python
x_padded = pad_sequences(x)
x_train1 , x_test1 , y_train1 , y_test1 = train_test_split(x_padded , y , test_size = 0.20 , random_state = 42)
```

```python
# Create Simple RNN

from keras.layers import Reshape

input_shape = (2500,)

model = Sequential()

model.add(Reshape((2500,1)))                                # Input Layer

model.add(SimpleRNN(units = 50 , return_sequences = False))  # Hidden Layer

model.add(Dense(units = 1 , activation = "sigmoid"))        # Output Layer
```

```python
# Compile

model.compile(optimizer = "adam" , loss = "binary_crossentropy" , metrics = ["accuracy"])
```

```python
# Train




model.fit(x_train1 , y_train1 , epochs = 5 , batch_size = 30 , validation_data = (x_test1 , y_test1))
```

```python
loss, accuracy = model.evaluate(x_test1 , y_test1)
print(f'Test Accuracy: ' , {accuracy})
```

```python
y_pred2 = model.predict(x_test1)
y_pred2
```

```python
y_pred_threshold_1 = (y_pred2 > 0.5).astype(int)
y_pred_threshold_1
```

```python
# Calculating the Accuracy

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test1 , y_pred_threshold_1)

print("Accuracy: ", accuracy)
```
